```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
from sklearn.cluster import KMeans
```

```python
sns.set()
```

```python
#load the data
data=pd.read_csv("C:\\Users\\india\\OneDrive\\Desktop\\data science\\archive\\3.01.Countryclusters.csv")
data
```

```python
#plotting the dataset
plt.scatter(data['Latitude'],data['Longitude'])
plt.xlim(-180,180)
plt.ylim(-90,90)
plt.show()
```

```python
#selecting the feature
x=data.iloc[:,1:3]
```

```python
#clustering
kmeans=KMeans(5)
kmeans.fit(x)
#clustering result
identified_clusters=kmeans.fit_predict(x)
identified_clusters
data_with_clusters=data.copy()
data_with_clusters['Clusters']=identified_clusters
plt.scatter(data_with_clusters['Latitude'],data_with_clusters['Longitude'],c=data_with_clusters['Clusters'],cmap='rainbow')
plt.show()
```

<h2>Write a python program to demonstrate K-Means clustering from scratch for K=2</h2>
<h3>Hint:use E.D and recalculate the centroid</h3>
<h4>sqrt()</h4>

```python
import math
data=[[10,2],[12,4],[16,6],[20,8],[24,10],[26,12],[30,14],[34,16]]
n=len(data)
print(n)
k1=[[10,2]]
k2=[[12,4]]
print(k1)
print(k2)
for i in range(2,n):
    ED1=math.sqrt(pow((data[i][0]-data[0][0]),2))+pow((data[i][1]-data[0][1]),2)
    ED2=math.sqrt(pow((data[i][0]-data[1][0]),2)+pow((data[i][1]-data[1][1]),2))
    if ED1>=ED2:
        k2.append(data[i])
    else:
        k1.append(data[i])
print(k1)
print(k2)
```
