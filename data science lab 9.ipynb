# WRITE A PYTHON PROGRAM FOR LINEAR, POLYNOMIAL AND MULTIPLE REGRESSION FOR A SPECIFIC DATA SET

```python
# LINEAR REGRESSION ---> Y=MX+C
# DATA POINTS : X(INDEPENDENT) AND Y (DEPENDENT)
# DRAW A BEST FIT LINE
import scipy
import matplotlib.pyplot as plt
from scipy import stats
```

```python
# LINEAR REGRESSION = Y=MX+C---> M(slope), C(intercept), X(ind. var)
x=[5,7,8,7,2,17,18,14,9,6]
y=[99,95,52,83,73,67,89,49,79,88]
#X AND Y ARE HISTORICAL DATA (if x is 5 then y is 99)
```

```python
slope,intercept,r,p, std_err=stats.linregress(x,y)
def myfunc(x):
    return slope*x+intercept #---> mx+c
z=map(myfunc,x)# it works just like a loop map()
print(z)
mymodel=list(z)
print(mymodel)
plt.scatter(x,y)
plt.plot(x,mymodel)
plt.show()
```

```python
#Polynomial regression
import numpy
x=[1,2,3,4,6,7,8,9,11,14,15,17,20,30,26]
y=[100,90,80,85,95,77,95,44,80,55,60,85,51,88,99]
```

```python
from sklearn.metrics import r2_score
p=numpy.polyfit(x,y,3)
mymodel=numpy.poly1d(p)
#find---> r(squared value)---0 and 1
#0 means ---> no relationship
#1 means--->perfect relationship
print(r2_score(y,mymodel(x)))# r squared value
myline=numpy.linspace(1,22,100)
plt.scatter(x,y)
plt.plot(myline,mymodel(myline))
plt.show()
```
