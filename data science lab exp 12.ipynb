EXP 12: Write a python program for prediction based on SVM(Support Vector Machine) classifier model.

```python
#library import to read the csv file
from pandas import read_csv
#TRAIN- TRAIN SPLIT---> Divide the data into training and testing 
from sklearn.model_selection import train_test_split
#To generate the classification report
from sklearn.metrics import classification_report
#To validate the model--> use confusion matrix
from sklearn.metrics import confusion_matrix
#Generate the accuracy score
from sklearn.metrics import accuracy_score
#import SVM
from sklearn.svm import SVC
```

```python
path="C:\\Users\\india\\OneDrive\\Desktop\\2nd year\\DATA SCIENCE LAB\\iris.csv"
names=['sepal_length','sepal_width','petal_length','petal_width','class']
dataset=read_csv(path,names=names)
```

```python
print(dataset)
```

```python
#convert this dataset into an array
array=dataset.values
#print(array)
#X---> independent values
X=array[:,0:4] #range--->[0,1,2,3]
#Y---> dependent values
Y=array[:,4]
X_train,X_validation,Y_train,Y_validation=train_test_split(X,Y,test_size=0.20,random_state=1)
```

```python
#step 1: Import the dataset
#step 2: convert it into array
#step 3: seperate the data into X (independent) y(dependent)
#X---> all the numerical values---> S_L,S_W,P_L,P_W---[0,1,2,3]
#Y---> classes----- 4 classes----[4]
#define X and Y
#X [select all the values in array and select from 0 to 3]
```

```python
#make predicitions
model=SVC (gamma='auto')
#Gamma controls your decision boundary
model.fit(X_train, Y_train)
predictions=model.predict(X_validation)
#evaluate the model
print(accuracy_score(Y_validation,predictions))
print(accuracy_score(Y_validation,predictions))
print(classification_report(Y_validation,predictions))
```
