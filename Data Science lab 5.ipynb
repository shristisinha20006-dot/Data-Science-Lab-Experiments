```python
# convert the dataframe to string 
new_df=(df.to_string())
print(new_df)
```

```python
#drop the NaN values
#variable_name=dataframe.dropna()
new_df=df.dropna()
print(new_df)
```

```python
#change the original file
#Inplace---deletes---->from original data
df.dropna(inplace=True)
print(df)
```

```python
# verify the data
df.loc[7,'Duration']
```

```python
# reallocate the value with the new one
df.loc[7,'Duration']=45
print(df)
```

```python
# rules 
#it will see every row and column and store the value at ...
#variable x
for x in df.index:
     if df.loc[x,"Duration"]>120:
      df.loc[x,"Duration"] =120
print(df)
```

```python
#Rules for deletion
for x in df.index:
    if df.loc[x,"Duration"]>120:
        df.drop(x,inplace=True)
```

```python
# drop the duplicate value
#variable name= dataframe(df)--> duplicated()---function---check---Duplicates---return---Boolean value
nv=df.duplicated()
print(nv)
#checking where the duplicate value lies??---nv---becomes---true
new_out=df.loc[nv==True]
print(new_out)
#drop_duplicate---function---drop the duplicate 
df.drop_duplicates(inplace=True)
print(df.to_string())
```

```python
import pandas as pd
import numpy as np
# creating a dictionary--->{key:values}---key--F.S----Value-->
dict={'First Score':[100,90,np.nan,95],
      'Second Score':[np.nan,45,56,np.nan],
      'Third Score':[np.nan,40,80,95]
     }
df=pd.DataFrame(dict)
print(df.to_string())
```

```python
# check if the value is null or not
#is null--->True--Value--NULL
print(df.isnull())
```

```python
#check if the value is not null
#notnull-->True--Value-->Notnull
print(df.notnull())
```

```python
print(df)
```

```python
#filling up the missing values
print(df.replace(to_replace=np.nan,value=40))
```

```python
#replace the nan values with mean
df_mean=df.mean()
print(df_mean)
#fillna--function--->fill the NaN--->user choice of value
data_new=df.fillna(df.mean())
print(data_new)
```
